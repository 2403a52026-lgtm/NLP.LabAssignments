```python
# spaCy for classical NLP pipeline and NER
import spacy
from spacy import displacy

# Hugging Face for transformer-based NER
from transformers import pipeline, AutoTokenizer, AutoModelForTokenClassification

# PyTorch backend
import torch

# Data handling
import pandas as pd
```

```python
# Load English small pretrained model
nlp = spacy.load("en_core_web_sm")
```

```python
sentences = [
    "Apple CEO Tim Cook announced new AI features in California.",
    "Lionel Messi scored a goal for Inter Miami in New York.",
    "The Prime Minister of India Narendra Modi visited France.",
    "Microsoft acquired OpenAI in a billion-dollar deal.",
    "NASA launched the Artemis mission from Florida."
]
```

```python
print("===== spaCy NER Results =====\n")

spacy_results = []

for sentence in sentences:
    doc = nlp(sentence)
    print(f"\nSentence: {sentence}")

    for ent in doc.ents:
        print(f"Entity: {ent.text} | Label: {ent.label_}")
        spacy_results.append([sentence, ent.text, ent.label_])
```

```python
spacy_df = pd.DataFrame(spacy_results, columns=["Sentence", "Entity", "Label"])
spacy_df
```

```python
displacy.render(nlp(sentences[0]), style="ent", jupyter=True)
```

```python
model_name = "dbmdz/bert-large-cased-finetuned-conll03-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)

ner_pipeline = pipeline("ner", model=model, tokenizer=tokenizer)
```

```python
print("\n===== Hugging Face NER Results =====\n")

hf_results = []

for sentence in sentences:
    results = ner_pipeline(sentence)

    print(f"\nSentence: {sentence}")

    for entity in results:
        word = entity["word"]
        label = entity["entity"]
        score = round(entity["score"], 4)

        print(f"Word: {word} | Label: {label} | Confidence: {score}")

        hf_results.append([sentence, word, label, score])
```

```python
cleaned_results = []

for sentence in sentences:
    results = ner_pipeline(sentence)

    current_entity = ""
    current_label = ""
    current_score = 0

    for entity in results:
        word = entity["word"]

        # Remove subword token prefix ##
        if word.startswith("##"):
            current_entity += word[2:]
        else:
            if current_entity:
                cleaned_results.append([sentence, current_entity, current_label, current_score])
            current_entity = word
            current_label = entity["entity"]
            current_score = round(entity["score"], 4)

    if current_entity:
        cleaned_results.append([sentence, current_entity, current_label, current_score])

hf_df = pd.DataFrame(cleaned_results,
                     columns=["Sentence", "Entity", "Label", "Confidence Score"])

hf_df
```
