```python
import chromadb
from sentence_transformers import SentenceTransformer
```

```python
client = chromadb.Client()

collection = client.create_collection(name="rag_collection")
print("Collection created successfully!")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded!")
```

```python
documents = [
    "Artificial Intelligence is the simulation of human intelligence.",
    "Machine Learning is a subset of AI.",
    "Deep Learning uses neural networks.",
    "Natural Language Processing deals with text data.",
    "RAG combines retrieval and generation.",
    "Neural networks are inspired by the human brain.",
    "Supervised learning uses labeled data.",
    "Unsupervised learning finds patterns in data.",
    "Reinforcement learning uses rewards and punishments.",
    "AI is widely used in healthcare and finance."
]

ids = [f"doc{i+1}" for i in range(len(documents))]

print("Dataset prepared!")
```

```python
embeddings = model.encode(documents)
print("Embeddings generated!")
```

```python
collection.add(
    documents=documents,
    embeddings=embeddings,
    ids=ids
)

print("Data stored in ChromaDB!")
```

```python
query = "What is Artificial Intelligence?"

query_embedding = model.encode([query])

results = collection.query(
    query_embeddings=query_embedding,
    n_results=3
)

print("Raw Results:\n", results)
```

```python
print("\nTop Retrieved Documents:\n")

for doc in results['documents'][0]:
    print("-", doc)
```

```python
def retrieve(query):
    query_embedding = model.encode([query])
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=2
    )
    return results['documents'][0]

# Test function
print("\nCustom Retrieval Test:\n")
print(retrieve("Explain machine learning"))
```

```python
queries = [
    "What is AI?",
    "Explain deep learning",
    "What is NLP?",
    "How does RAG work?"
]

for q in queries:
    print(f"\nQuery: {q}")
    print("Results:", retrieve(q))
```
